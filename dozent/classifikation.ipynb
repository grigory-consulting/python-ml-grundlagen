```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

X, y = load_iris(return_X_y=True)          # 150 Blüten, 4 Merkmale
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)
model = KNeighborsClassifier(n_neighbors=5) # Klasse/Zielgröße wird anhand von Nachbarn bestimmt 

model.fit(X_train, y_train)                # lernen

# Vorhersage
print(model.predict(X_test[:5]))           # -> [0 1 1 0 2]

# Echte Daten 
print(y_test[:5])                          # -> [0 1 1 0 2]
print(model.score(X_test, y_test))         # Genauigkeit 

```

```python
model.kneighbors(X_test)
```
