```python
# Verschachtelung

alter = 20

if alter >= 18:
    if alter >= 21:
        print("volljährig und mindestens 21")
    else:
        print("volljährig, aber unter 21")
else:
    print("minderjährig")

# DONE flache Fassung mit elif : jede Bedingung steht auf derselben Ebene
if alter >= 18 and alter < 21:
    print("volljährig, aber unter 21") 
elif alter >= 21:
    print("volljährig und mindestens 21")
else:
    print("minderjährig") 


# Bei Einzeilern 
if 18 <= alter < 21: print("volljährig, aber unter 21") 
elif alter >= 21: print("volljährig und mindestens 21")
else: print("minderjährig")
```

```python
def jahresbeitrag(monatsbeitrag): # def ... defininiert die Funktion, in der Klammer die Parameter  
    ergebnis = 12*monatsbeitrag
    return ergebnis # Zurückgabe von Werten (optional), ohne return gibt es None. 

jahresbeitrag(412.5)
jahresbeitrag(398.5) + jahresbeitrag(412.5)
```
