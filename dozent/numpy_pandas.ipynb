```python
# Zufallszahlengenerator 

rng = np.random.default_rng(1) # Seed festlegen für Reproduzierbarkeit
heights = rng.normal(170, 10, 1000) # 1000 Werte, normalverteilt mit mu = 170 und sigma = 10

# Maske = boolesche Arrays 
maske = heights > 190
maske.sum() # Anzahl der Großen
maske.mean() # Mittwert der Maske = Anteil
heights[maske] # Alle Elemente, die die Maske erfüllen
heights[~maske] # Alle Elemente, die die Maske nicht erfüllen
```

```python
import pandas as pd

df = pd.read_csv("data/versicherte.csv") # df ... DataFrame 
#df.info()
df.tail(7) # head ... die ersten, tail ... die letzten
```

```python
df.alter # Spaltenbasierter Zugriff 
df["alter"] # dasselbe 

# Wichtige Zugriff: loc (namentlich) und iloc (indexbasiert)
spalten = ["vorname", "nachname", "stadt"]
mit_loc = df.loc[10:20, spalten]
mit_loc
```

```python
# dasselbe mit iloc 
mit_iloc = df.iloc[10:21, [1,2,7]] # indexbasiert angeben, Ende 21 gehört nicht dazu
mit_iloc
```

```python
# mit Maske filtern 
# männliche Raucher mit BMI > 30 
# in numpy und pandas & ... für und, | ... für oder, ~ ... für nicht
maske = (df["geschlecht"] == "männlich") &  (df["bmi"]>30) & (df["raucher"])
df[maske]

# mit Query (SQL-ählich oder SQL-like)
query = df.query("geschlecht == 'männlich' and bmi > 30 and raucher")
query
```

```python
# Altersgruppen und bmi_klasse mit pd.cut

df["altersgruppe"] = pd.cut(df["alter"], bins= [0, 18, 65, 120],
                           labels=["Kind", "Erwachsen", "Senior"], right=False)
# right = False -> rechte Grenze gehört nicht dazu

# TODO BMI 
# unter 18.5 -> Untergewicht
# >=18.5 aber unter 25 -> Normalgewicht
# >= 25 aber unter 30 -> Übergewicht
# >= 30 -> Adipositas 
df["bmi_klasse"] = pd.cut(df["bmi"], bins=[0, 18.5, 25, 30, 100],
                          labels=["Untergewicht", "Normalgewicht", "Übergewicht", "Adipositas"],
                          right=False)


df["altersgruppe"].value_counts() # Zählen 
df["bmi_klasse"].value_counts(dropna=False)
```
