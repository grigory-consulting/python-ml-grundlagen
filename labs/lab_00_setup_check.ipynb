## Block 3: Erster Plot als Funktionstest

8. Führen Sie die Zelle zu Aufgabe 8 unverändert aus. Sie zeichnet eine Sinuskurve und testet damit NumPy und Matplotlib zusammen. **Erwartet:** eine Wellenlinie zwischen -1 und 1 mit einer vollen Schwingung, Achsen beschriftet mit „x" und „sin(x)". Ausgabe darunter: `100 Punkte, größter Wert 1.0`.
9. Wandeln Sie den Plot ab: Zeichnen Sie zusätzlich den Kosinus in dasselbe Diagramm, vergeben Sie für beide Kurven ein `label` und blenden Sie die Legende ein. **Erwartet:** zwei Kurven, die Kosinuskurve beginnt bei 1, die Sinuskurve bei 0. Ausgabe darunter: `Kurven im Diagramm: 2`.

```python
# Aufgabe 8: Sinuskurve zeichnen (unverändert ausführen)
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)     # 100 Werte von 0 bis 2 Pi
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.set_title("Funktionstest: Sinus")
plt.show()

print(len(x), "Punkte, größter Wert", round(float(y.max()), 2))
```

```python
# Aufgabe 9: Sinus und Kosinus in einem Diagramm
# Tipp: np.cos(x), ax.plot(x, y, label="..."), ax.legend()
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), label="sin(x)")
# Hier die Kosinuskurve ergänzen

ax.set_xlabel("x")
ax.set_ylabel("Funktionswert")
# Hier die Legende einblenden
plt.show()

print("Kurven im Diagramm:", len(ax.lines))
```
